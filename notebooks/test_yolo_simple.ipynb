```python
from ultralytics import YOLO
from PIL import Image
import matplotlib.pyplot as plt
from collections import Counter
import numpy as np
```

```python
# Chemin de l'image à analyser
image_path = "media_1770629400.jpg"  

# Afficher l'image
image = Image.open(image_path)
plt.figure(figsize=(15, 10))
plt.imshow(image)
plt.axis('off')
plt.title("Place de la Comédie - Image originale")
plt.show()

print(f"Taille: {image.size}")
```

```python
# Charger les modèles YOLOv8

model1 = YOLO('yolov8n.pt')
print("✓ YOLOv8n chargé")
model2 = YOLO('yolov8s.pt')
print("✓ YOLOv8s chargé")
model3 = YOLO('yolov8m.pt')
print("✓ YOLOv8m chargé")
model4 = YOLO('yolov8l.pt')
print("✓ YOLOv8l chargé")
model5 = YOLO('yolov8x.pt')
print("✓ YOLOv8x chargé")

models = [
    ("YOLOv8n", model1),
    ("YOLOv8s", model2),
    ("YOLOv8m", model3),
    ("YOLOv8l", model4),
    ("YOLOv8x", model5)
]
```

```python
# Test de chaque modèle sur l'image
for model_name, model in models:
    print(f"Test du modèle {model_name}...")
    results = model(image_path, conf=0.25)
    for result in results:
        im_array = result.plot()
        im = Image.fromarray(im_array[..., ::-1])
        plt.figure(figsize=(18, 12))
        plt.imshow(im)
        plt.axis('off')
        plt.title(f"Place de la Comédie - Détections {model_name}", fontsize=16)
        plt.show()
```
